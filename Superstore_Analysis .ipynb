114. Monthly Sales Trend

## Step 114: Monthly Sales Trend
Monthly sales are visualized to identify seasonal patterns,fluctuations, and changes in revenue over time.

```python
monthly_sales=df.groupby(
df["Order Date"].dt.to_period("M")
)["Sales"].sum().reset_index()
monthly_sales["Order Date"]=monthly_sales["Order Date"].astype(str)
plt.figure(figsize=(15,6))
sns.lineplot(
    data=monthly_sales,
    x="Order Date",
    y="Sales",
    marker="o"
)
plt.title("Monthly Sales Trend",
          fontsize=16,fontweight="bold")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=60,ha="right")
plt.tight_layout()
plt.show()
```

### Business Insight
Monthly sales show an overall upward trend with periodic fluctuations,indicating business growth over time while also suggesting seasonal variations in customer demand.

## Step 125: Correlation Heatmap  
The Correlation Heatmap shows the strength and direction of relationships between numerical variables such as Sales,Quantity,Discount,Profit,Profit Margin, and Shipping Days.It helps identify variables that may have meaningful relationships with each other.
